```python
# spacy implementation

import spacy
nlp = spacy.load('en_core_web_sm')

def count_words_spacy(text):
  doc = nlp(text)
  words = [token.text for token in doc if not token.is_punct]
  return len(words)
text="Let's explore nlp with python."
doc = nlp(text)
print([token.text for token in doc])
print(count_words_spacy(text))
```

```python
from collections import Counter

def word_frequency(text):
  doc = nlp(text)
  words = [token.text.lower()  for token in doc if not token.is_punct]
  return Counter(words)

text = "NLP is fun, NLP is powerful, and NLP is amazing in the future."
print(word_frequency(text))
```

## NLP - Task 1
### Short vs Long sentences

```python
from nltk.tokenize import sent_tokenize

def categorize_sentences(text):
  sentences = sent_tokenize(text)
  categorized = {"short":[], "long":[]}

  for sentence in sentences:
    word_count = len(word_tokenize(sentence))
    if word_count < 5:
      categorized["short"].append(sentence)
    else:
      categorized["long"].append(sentence)

  return categorized
```

```python
text = "Hello, NLP is fun. I like it very much. I hope everyone is following this lecture. Hello, Guys."

categories= categorize_sentences(text)
print(categories)
print("Short Sentences:", categories['short'])
print("Long Sentences:", categories['long'])
```

```python
## Question


def quest_categ(text):
  sentences = sent_tokenize(text)
  categorized = {"Question":[], "Not a Question":[]}

  for sentence in sentences:
    if sentence.strip().endswith('?'):
      categorized["Question"].append(sentence)
    else:
      categorized["Not a Question"].append(sentence)

  return categorized
```

```python
text = "who are you?"

categories= quest_categ(text)
print(categories)
print("Question:", categories['Question'])
print("Not a Question:", categories['Not a Question'])
```

```python
### Rule Based sentiment analysis

import time

positive_words= {"happy", "sad", "excellent", "love", "amazing", 'win', 'prize' }
negative_words= {"sad", "bad", "worse", "terrible", "hate"}
```

```python
def sentiment_analy(text):
  start_time = time.time()
  words = text.lower().split()

  pos_count = sum(1 for word in words if word in positive_words)
  neg_count = sum(1 for word in words if word in negative_words)

  sentiment = "Positive Sentiment" if pos_count > neg_count else "Negative Sentiment" if neg_count > pos_count else "Neutral Sentiment"

  end_time = time.time()
  execution_time = end_time - start_time
  return sentiment, execution_time
```

```python
text = "I am very sad"
sentiment, exec_time = sentiment_analy(text)
print("Sentiment:", sentiment)
print("Execution Time:", exec_time, "seconds")
```

```python
text = "I am in a happy mood, but i feel sad about my friend who had a terrible fall from the table"
sentiment, exec_time = sentiment_analy(text)
print("Sentiment:", sentiment)
print("Execution Time:", exec_time, "seconds")
```
