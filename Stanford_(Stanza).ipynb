```python
import stanza
stanza.download('en')
nlp=stanza.Pipeline('en',processors='tokenize,mwt,lemma,pos,depparse,ner,constituency')
```

```python
# Event Extraction

text = 'jhon bought a new laptop from Amazon'
doc = nlp(text)
for sent in doc.sentences:
  for word in sent.words:
    if word.upos in ['VERB', 'AUX']:
      print(f"Event:{word.text}")

      for dep in sent.words:
        if dep.head == word.id:
          print(f" Argument({dep.deprel}):{dep.text}")
```

```python
# Temporal information extraction

text = 'jhon bought a new laptop from Amazon at 27 march, 2024 at 10 AM.'
doc = nlp(text)

for ent in doc.ents:
  if ent.type == 'DATE' or ent.type == 'TIME':
    print(f"Temporal Expression: {ent.text} --> {ent.type}")
```

```python
## Question answering pipelines

question = "who helped India in 1947"
context = "Mahatma Gandhi helped India get the independence in 1947"

doc = nlp(context)

for sent in doc.sentences:
  for word in sent.words:
    if word.deprel == 'nsubj':
      print(f"Possible answer: {word.text}")
```

```python
# Structure aware text summarization
text = " The Manager, who was extremely busy, scheduled a meeting with the team "
doc = nlp(text)

for sent in doc.sentences:
  essential = [word.text for word in sent.words if word.deprel in ['nsubj','obj','root']]
  print(" ".join(essential))
```

```python
# Multiliguanal document analysis
stanza.download('hi')
hi_nlp = stanza.Pipeline(lang='hi')
```

```python
text = "राम ने सीता के लिए फूल खरीदे।"
doc = hi_nlp(text)
for sent in doc.sentences:
  for word in sent.words:
      print(f"{word.text} ({word.upos} - {word.deprel} --> {word.head})")
```

```python
text1 = "The cat sat on the mat."
text2 = "A feline was sitting on the rug. "

doc1 = nlp(text1)
doc2 = nlp(text2)

def get_dependencies(doc):
  return set((w.text, w.deprel) for s in doc.sentences for w in s.words)

print(get_dependencies(doc1) & get_dependencies(doc2))
```

```python
# Named entity normalizaion

text = "Apple was founded by Steve Jobs in Cupertino"

doc = nlp(text)

for ent in doc.ents:
  print(f"Entity: {ent.text} : {ent.type}")
```

```python
# Syntax aware machine translation

text = "the dog chased the cat under the table"

doc = nlp(text)

for sent in doc.sentences:
  print(sent.constituency)
```
