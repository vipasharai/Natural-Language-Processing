```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download('punkt_tab')
```

```python
text ="Hello World! NLP is amazing. Let's tokenize this sentence"
print("word tokenize:",word_tokenize(text))
print("sentence tokenize:",sent_tokenize(text))
```

```python
print("Len word tokenize:",len(word_tokenize(text)))
print("len sentence tokenize:",len(sent_tokenize(text)))
```

```python
# using spacy

import spacy
nlp = spacy.load('en_core_web_sm')
doc = nlp(text)
print("word tokenize:",[token.text for token in doc])
print("sentence tokenize:",[sent.text for sent in doc.sents])
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
import contractions

stop_words = set(stopwords.words('english'))
text = contractions.fix(text)
doc = nlp(text)
words = [token.text for token in doc if not token.is_punct]
filtered_words = [word for word in words if word not in stop_words]

filtered_freq = Counter(filtered_words)

print(len(filtered_words))
print(filtered_freq)
```

```python
## NER (Name entity recognitzation) with context Analysis

import spacy
from collections import Counter

text = "Apple is looking at buying a UK startup for 1 billion USD."
doc = nlp(text)

entity_freq = Counter([ent.text for ent in doc.ents])

for ent, freq in entity_freq.most_common():
    print(f"Entity :{ent},Count:{freq},Label:{nlp(ent)[0].ent_type_}")

```

```python
### N- Grams with frquency analysis

from nltk.util import ngrams

bigrams = list(ngrams(words,2))
trigrams = list(ngrams(words,3))

bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)
```

```python
print(bigram_freq.most_common(5))
print(trigram_freq.most_common(5))
```

```python
## KEyword extraction using TF_IDF with tokenization

from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform([text])

feature_array =  vectorizer.get_feature_names_out()

importance = X.toarray().flatten()
importance_words = sorted(zip(feature_array,importance),key=lambda x:x[1],reverse=True)[:5]
print([word for word, _ in importance_words])
```

```python
# TEXT Summarization
```

```python
text = """
This is a string tokenization lecture. we are on a way to learn NLp
effectively. we hope that we can excel in our career after learning NLp
"""

doc = nlp(text)
words = [token.text for token in doc if token.is_alpha and token.text not in stopwords.words("english")]

filtered_freq = Counter(words)


sentences_scores = {}
for sent in doc.sents:
  for word in word_tokenize(sent.text.lower()):
    if word in filtered_freq:
      sentences_scores[sent.text] = sentences_scores.get(sent.text,0)+filtered_freq[word]
sorted_sentences = sorted(sentences_scores ,key=sentences_scores.get,reverse=True)
summary = " ".join(sorted_sentences[:2])
print(summary)
```
