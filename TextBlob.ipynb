## Text Summarization

```python
from collections import Counter

text = """Python is a great programming language.
It is widely used for data analysis, machine learning, web development, and automation."""

blob = TextBlob(text)
nouns = [word.lower() for word, tag in blob.tags if tag == "NN"]

common_nouns = Counter(nouns).most_common(5)
summary = " ".join([word for word, _ in common_nouns])

print(summary)
```

## Keyword Extract

```python
blob = TextBlob("Python is a great programming language. It is widely used for data analysis, machine learning, web development, and automation.")

noun_phrases = blob.noun_phrases

adjactives = [word for word, tag in blob.tags if tag=='JJ']

verbs = [word for word, tag in blob.tags if tag.startswith('VB')]

keywords = set(noun_phrases + adjactives + verbs)

print("Noun phrases:", noun_phrases)
print("Adjectives:", adjactives)
print("Verbs:", verbs)
print("keywords:", keywords)
```

## Topic Clustering

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

docs = [
    "Artificial intelligence is changing the world.",
    "Machine learning helps computers learn from data.",
    "Natural language processing analyzes human language.",
    "Data science helps businesses make better decisions.",
    "Python programming is easy to learn.",
    "Deep learning uses neural networks.",
    "The Indian education system needs improvement.",
    "Social media connects people around the world.",
    "Climate change affects the global environment.",
    "Cyber security protects computer networks."
]
cleaned_docs = [' '.join(TextBlob(doc).words.lower()) for doc in docs]

vectorizer = TfidfVectorizer()
x = vectorizer.fit_transform(cleaned_docs)

kmeans = KMeans(n_clusters=3, random_state=42).fit(x)

for i, label in enumerate(kmeans.labels_):
  print(f"Document {i+1} belongs to cluster {label+1}")
```
