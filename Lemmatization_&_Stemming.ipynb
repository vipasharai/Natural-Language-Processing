```python
import nltk
import spacy
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from collections import Counter
```

```python
# Download weruired resources
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# Initialize the porters and stemmers

porter = PorterStemmer()
snowball = nltk.SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()
nlp = spacy.load("en_core_web_sm")
```

```python
corpus = {
    "news": "The government announced a new education policy today. The policy aims to improve the quality of education and provide better opportunities for students.",

    "wikipedia": "Artificial intelligence is a branch of computer science that focuses on creating machines capable of performing tasks that normally require human intelligence.",

    "science_paper": "Machine learning is a subfield of artificial intelligence that enables computers to learn patterns from data. Deep learning uses neural networks to solve complex problems."
}

```

```python
tokenize_corpus = {key: nltk.word_tokenize(value) for key, value in corpus.items()}
tokenize_corpus
```

```python
def apply_stemming(text_tokens, stemmer):
  return [stemmer.stem(word) for word in text_tokens]

for domain, tokens in tokenize_corpus.items():
  print(domain.upper())
  print(apply_stemming(tokens, porter))
  print(apply_stemming(tokens, snowball))
```

 ## Applying Lemmatization

```python
def apply_lemmatization(text_tokens, lemmatizer, pos = wordnet.VERB):
  return [lemmatizer.lemmatize(word, pos) for word in text_tokens]

for domain, tokens in tokenize_corpus.items():
  print(domain.upper())
  print(apply_lemmatization(tokens, lemmatizer))
```

## Using Spacy

```python
def spacy_lemmatization(text):
  doc = nlp(text)
  return [token.lemma_ for token in doc]

for domain, text in corpus.items():
  print(domain.upper())
  print(spacy_lemmatization(text))
```

## NLP Task

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
GEN_Z_DICT = {
    "money": "bread", "work": "grind", "friend": "bestie", "tired": "exhausted af",
    "smart": "big-brain", "cool": "vibe check passed", "boring": "mid",
    "excited": "hyped", "love": "obsessed", "understand": "get it, bestie",
    "bad": "sus", "good": "fire", "funny": "lowkey hilarious", "amazing": "goated",
}

def normal_mode(text):
  """returns the text in normal form after lemmatization"""
  return " ".join(lemmatizer.lemmatize(w) for w in word_tokenize(text))

def gen_z_mode(text):
  words = word_tokenize(text.lower())
  return " ".join([GEN_Z_DICT.get(w, w) for w in words])

def translate():
  text = input("Enter a message")
  print(normal_mode(text))
  print(gen_z_mode(text))
```

```python
translate()
```
