```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
pos_sent = "NLP is Amazing"
ner_sent = "Google was invented by Larry Page and Sergey Brin"
```

```python
# POS Tagging
pos_doc = nlp(pos_sent)
pos_tags = [(token.text, token.pos_) for token in pos_doc] #[(token.text, token.pos) for token in pos_doc]
print(pos_tags)
```

```python
# NER - Name Entity Recognition
ner_doc = nlp(ner_sent)
ner_tags = [(ent.text, ent.label_) for ent in ner_doc.ents]
print(ner_tags)
```
