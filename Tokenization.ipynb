# Tokenization

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize

# Sample
text_word = "I love programming"
sent_word = "i love programming, and i also love python. It is my passion"
token_words = word_tokenize(text_word)
token_sent = sent_tokenize(sent_word)
```

```python
print(token_sent)
```

```python
for i in token_sent:
  x = word_tokenize(i)
  print(x)
```

## StopWord Removal

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))  # french ,...
print(len(stop_words))
print(stop_words)
```

```python
text = " i haven't thought this through by doing most of these against themselves"
len(text)
```

```python
tokens = word_tokenize(text)
print(tokens)
print(len(tokens))
```

```python
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
```

```python
print(filtered_tokens)
```

## Remove Puncuation

## Stemming

```python
from nltk.stem import PorterStemmer

# sample words
words = ['runner', 'burnt', 'happily', 'easily']

stremmer = PorterStemmer()
stremmer = [stremmer.stem(word) for word in words]
print(stremmer)
```

```python
words = ['runs', 'runner', 'running']
stremmer = PorterStemmer()
stremmer = [stremmer.stem(word) for word in words]
print(stremmer)
```

## Lemmatization

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
words = ['runner', 'burnt', 'happily']
```

```python
lemmatized = [token.lemma_ for token in nlp(' '.join(words))]
print(lemmatized)
```

```python
words = "runner , burnt, flying"
```

```python
doc =nlp(words)
```

```python
lemmatized = [token.lemma_ for token in doc]
print(lemmatized)
```

```python
for token in doc:
  print(f"{token.text:<15}{token.pos_:<10}{token.lemma_}")
```
