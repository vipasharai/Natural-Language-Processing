# ***Gated Recurrent Unit***

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

## Sample dataset

```python
texts = ['I loved this movie!', 'It was a terrible movie']
labels = np.array([1,0])
```

```python
# Tokenize

tokenizer = Tokenizer(num_words = 10000)
tokenizer.fit_on_texts(texts)
X = tokenizer.texts_to_sequences(texts)
X = pad_sequences(X,padding = 'post')
```

## Build The model

```python
model = Sequential()
model.add(Embedding(input_dim = 10000, output_dim = 128, input_length = X.shape[1]))
model.add(GRU(units = 64))
model.add(Dense(units = 1, activation = 'sigmoid'))

# Compile the model
model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])

# Train the model
model.fit(X, labels, epochs = 5)
```

```python
# predict new model
new_review = ['I hated this movie!']
new_review_seq = tokenizer.texts_to_sequences(new_review)
new_review_seq = pad_sequences(new_review_seq, padding = 'post', maxlen = X.shape[1])
prediction = model.predict(new_review_seq)

print(prediction[0][0])
```

## NER

```python
# sample sentences and labels

texts = [['Barack', 'Obama', 'was', 'born', 'in', 'hawaii']]
labels = [['B-PER', 'I-PER', 'O', 'O', 'O', "B-LOC"]]

# Tokenize
tokenizer.fit_on_texts(texts)
x = tokenizer.texts_to_sequences(texts)
x = pad_sequences(x,padding = 'post')

# Label Mapping
label_map = {'O': 0,
             'B-PER': 1, 'I-PER': 2,
             'B-LOC': 3}
y = [[label_map[label] for label in sent] for sent in labels]
y = pad_sequences(y, padding= 'post')
y= np.expand_dims(y, -1)
```

```python
from tensorflow.keras.layers import Embedding, GRU, Dense, Dropout, TimeDistributed

model = Sequential()
model.add(Embedding(input_dim = len(tokenizer.word_index) + 1, output_dim=128, input_length = x.shape[1]))
model.add(GRU(units = 64, return_sequences = True))
model.add(Dropout(0.1))
model.add(TimeDistributed(Dense(units = len(label_map), activation = 'softmax')))
```

```python
# Compile the model
model.compile(optimizer = 'adam', loss = 'sparse_categorical_crossentropy', metrics = ['accuracy'])

# Train the model
model.fit(x, y, epochs = 10)
```

```python
# predict new model
new_review = ['Obama', 'visited', 'Germany']
new_review_seq = tokenizer.texts_to_sequences(new_review)
new_review_seq = pad_sequences(new_review_seq, padding = 'post', maxlen = X.shape[1])
prediction = model.predict(new_review_seq)

predicted_labels = [list(label_map.keys())[np.argmax(pred)] for pred in prediction[0][:len(new_review_seq)]]
print(list(zip(new_review, predicted_labels)))
```
